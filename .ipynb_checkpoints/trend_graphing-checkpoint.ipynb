```python
import pandas as pd
from matplotlib import pyplot as plt
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
df = pd.read_csv('data/multi_regression.csv', index_col = False)
df = df.dropna()
```

```python
#plot sentiment change vs trade balance change
fig = make_subplots(specs=[[{"secondary_y": True}]])

# Add Sentiment trace (primary y-axis)
fig.add_trace(go.Scatter(x = df['Month'], y = df['Sentiment_Change'], name = 'Sentiment Change',
            mode = 'lines+markers'), secondary_y = False)

# Add Exchange Rate trace (secondary y-axis)
fig.add_trace(go.Scatter(x = df['Month'], y = df['Log_trade_balance_change_lag'], name = 'Log Change in Trade Balance (Lagged)',
        mode = 'lines+markers'), secondary_y = True)

# Update layout
fig.update_layout(title_text = 'Change in Bilateral Sentiment vs Logarithmic Change in Trade Balance Over Time: ',hovermode = 'x unified')

# Set axis titles
fig.update_xaxes(title_text = 'Month')
fig.update_yaxes(title_text = 'Sentiment Change', secondary_y = False)
fig.update_yaxes(title_text = 'Log Change in Trade Balance', secondary_y = True)

# Show plot
fig.show()

#plot sentiment change vs exchange rate change
fig = make_subplots(specs=[[{"secondary_y": True}]])

# Add Sentiment trace (primary y-axis)
fig.add_trace(go.Scatter(x = df['Month'], y = df['Sentiment_Change'], name = 'Sentiment Change',
            mode = 'lines+markers'), secondary_y = False)

# Add Exchange Rate trace (secondary y-axis)
fig.add_trace(go.Scatter(x = df['Month'], y = df['Log_EXCHUS_Change_Lag'], name = 'Log Change in Exchange Rate (Lagged)',
        mode = 'lines+markers'), secondary_y = True)

# Update layout
fig.update_layout(title_text = 'Change in Bilateral Sentiment vs Logarithmic Change in Exchange Rate Over Time: ',hovermode = 'x unified')

# Set axis titles
fig.update_xaxes(title_text = 'Month')
fig.update_yaxes(title_text = 'Sentiment Change', secondary_y = False)
fig.update_yaxes(title_text = 'Log Change in Exchange Rate', secondary_y = True)

# Show plot
fig.show()
```
